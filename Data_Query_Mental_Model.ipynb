```python
import pandas as pd
```

**Task 1: Data Preparation and Missing Value Handling (30 points)**

```python
students_data = {
    'student_id': [101, 102, 103, 104, 105, 106, 107],
    'name': ['Alice', 'Bob', None, 'David', 'Emma', 'Frank', 'Grace'],
    'email': ['alice@email.com', 'bob@email.com', 'charlie@email.com', None, 'emma@email.com', 'frank@email.com', 'grace@email.com'],
    'city': ['Mumbai', 'Delhi', 'Bangalore', 'Mumbai', None, 'Chennai', 'Delhi']
} # students actual data

enrollments_data = {
    'student_id': [101, 102, 103, 105, 108, 109],
    'course_name': ['Python', 'Data Science', 'Python', 'Machine Learning', 'AI', 'Python'],
    'enrollment_date': ['2024-01-15', '2024-01-20', '2024-02-01', '2024-02-10', '2024-02-15', '2024-03-01']
} # student's enrollment actual data

scores_data = {
    'student_id': [101, 102, 104, 105, 106],
    'exam_score': [85, 92, 78, 88, 95]
} # student's score actual data

students_data_df = pd.DataFrame(students_data) # students data's dataframe using pandas
enrollments_data_df = pd.DataFrame(enrollments_data) # enrollment data's dataframe using pandas
scores_data_df = pd.DataFrame(scores_data) # score data's dataframe using pandas

# Starting Task 1
print("===== Task 1 : Data Preparation =====\n")
print(f"Original Student DataFrame: -\n{students_data_df}\n") # Displaying Original Student Data
print("\n----------------------------\n")

print("Null Value Analysis: -") # Counting Null Values In Student's Data Every Coulmn
for column in students_data_df.columns:
  null_count = students_data_df[column].isnull().sum() # counting null values in terms of numbers
  percentage = round((null_count / len(students_data_df)) * 100, 2) # counting null values in terms of percentage and rounding upto 2 digit
  print(f"Column : {column}, Null : {null_count} ({percentage}%)")
print("\n----------------------------\n")

# Filling Missing City's With Unknown Using fillna() Method
students_data_df['city'] = students_data_df['city'].fillna('Unknown')

# Removing Student Record Where Name Is Empty Or Not Found Using dropna() Method
students_data_df = students_data_df.dropna(subset=['name'])

cleaned_students_data_df = students_data_df # Storing Cleaned Data In A New Variable
print(f"\nCleaned Student DataFrame: -\n{cleaned_students_data_df}\n")
```

### Task 1 Explanation

In this task, three DataFrames were created using the provided student, enrollment, and score data.

The students DataFrame was analyzed for missing values by calculating the count and percentage of null values in each column.

Missing values in the `city` column were filled with "Unknown", and rows where the `name` column was missing were removed. The cleaned students DataFrame was then displayed.

**Task 2: Multiple Join Operations (40 points)**

```python
print("===== Task 2 : Join Operations =====\n")

# Inner Join
inner_join = pd.merge(cleaned_students_data_df[['student_id', 'name']], enrollments_data_df, on='student_id', how='inner') # Inner Join or Intesection
print(f"Inner Join Result ({len(inner_join)} rows) \n{inner_join}\n") # Counting Total students after inner join and printing inner join result

# Counting Students Who Were Not In Enrollment Table
print(f"Excluded Students: - {list(cleaned_students_data_df['student_id'][~cleaned_students_data_df['student_id'].isin(inner_join['student_id'])])} - Not In Enrollment Table")
print("\n----------------------------\n")

# Left Join
left_join = pd.merge(cleaned_students_data_df[['student_id', 'name']], enrollments_data_df, on='student_id', how='left') # Left Join
print(f"\nLeft Join Result ({len(left_join)} rows) \n{left_join}\n") # Counting Total students after left join and printing left join result

# Counting Students With Null Course Name
print(f"Students with null course_name: - {list(left_join[left_join['course_name'].isnull()]['student_id'])}")
print("\n----------------------------\n")

# Right Join
right_join = pd.merge(cleaned_students_data_df[['student_id', 'name']], enrollments_data_df, on='student_id', how='right') # Right Join
print(f"\nRight Join Result ({len(right_join)} rows) \n{right_join}\n") # Counting Total students after right join and printing right join result

# Counting Students With Null Name
print(f"Finding Student IDs Without Names: - {list(right_join[right_join['name'].isnull()]['student_id'])}")
print("\n----------------------------\n")

# Full Outer Join
outer_join = pd.merge(cleaned_students_data_df[['student_id', 'name']], enrollments_data_df, on='student_id', how='outer', indicator=True) # Full Outer Join
print(f"\nOuter Join Result ({len(outer_join)} rows) \n{outer_join}\n") # Counting Total students after outer join and printing outer join result

# Counting Students Present in Both Tables
print(f"Finding Student Where Student Name Or Course Name Is Null: - {list(outer_join[(outer_join['name'].isnull()) | (outer_join['course_name'].isnull())]['student_id'])}") # Printing result
```

Reason:

If a student does not exist in the enrollments table, the record becomes NaN.

### Right Join Analysis

Total rows in result: 6

Student IDs without names: [103, 108, 109]

Reason:
A right join keeps all records from the enrollments table.

### Outer Join Analysis

Rows where name or course_name is null represent records that exist in only one of the tables.

**Task 3: Lookup Operation and Automation**

```python
print("===== Task 3 : Lookup Operation and Automation =====\n")

# 1) Lookup Operation
lookup_dict = dict(zip(scores_data_df['student_id'], scores_data_df['exam_score'])) # Creating Lookup Dictionary

# Create a copy of the cleaned students DataFrame
students_lookup = cleaned_students_data_df.copy()

# Use .map() to add exam scores to the students DataFrame based on matching student_id
students_lookup['exam_score'] = students_lookup['student_id'].map(lookup_dict)

# Display the students with their mapped exam scores (NaN for students without scores)
print("Lookup Operation Result:")
print(students_lookup[['student_id','name','exam_score']]) # Including only id, name, and score column for better readability
print("\n----------------------------\n")

# 2) Performance Comparison
students_scores_merge = pd.merge(cleaned_students_data_df, scores_data_df, on='student_id', how='left') # left join
print(f"\nLeft Join Result: -\n{students_scores_merge[['student_id','name','exam_score']]}\n") # Printing Left Join Results
print("\n----------------------------\n")

# 3) Simple Automation

def auto_merge(df1, df2, join_type, key_column):

  merged_df = pd.merge(df1, df2, on=key_column, how=join_type) # Performs Mege Or Join

  return {'result_df': merged_df,'row_count': len(merged_df),'join_type': join_type} # Returning The Dictionary As Asked


test_left = auto_merge(cleaned_students_data_df, scores_data_df, 'left', 'student_id') # Storing Function's Result Left Join
# Printing Formatted Result
print("Automation Function Test: -")
print(f"\nJoin Type: {test_left['join_type']}") # Printing Join Type
print(f"Rows in Result: {test_left['row_count']}") # Printing Total Rows
print(test_left['result_df']) # Printing Actual Join Result
print("\n----------------------------\n")

test_inner = auto_merge(cleaned_students_data_df, enrollments_data_df,'inner','student_id') # Storing Function's Result Inner Join
# Printing Formatted Result
print("\nAutomation Function Test: -")
print(f"\nJoin Type: {test_inner['join_type']}") # Printing Join Type
print(f"Rows in Result: {test_inner['row_count']}") # Printing Total Rows
print(test_inner['result_df']) # Printing Actual Join Result
```

### Lookup Operation Explanation

A dictionary mapping student_id to exam_score is created using the scores DataFrame.  

The .map() function is then used to match each student_id in the students DataFrame with its corresponding exam score from the dictionary.  

If a student does not have a score in the scores DataFrame, the result shows NaN.

---

### Performance Comparison

The .map() function performs a direct lookup using a dictionary, so it is faster when adding a single column based on a key.

In contrast, pd.merge() combines entire DataFrames, which requires more processing.

Therefore, .map() is more efficient for simple lookup tasks.

---

### Simple Automation

A reusable function auto_merge() is created to automate merge operations between two DataFrames.

The function takes two DataFrames, the join type, and the key column as inputs.

It performs the specified merge and returns a dictionary containing the merged DataFrame, the number of rows in the result, and the type of join used.

This helps simplify merge operations and allows the same function to be used with different join types.
